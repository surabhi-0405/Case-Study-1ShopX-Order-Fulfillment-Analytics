**Step 19.1 — PRIMARY KEY VALIDATION**

```python
# ============================================================
# PHASE 2 - STEP 19: PRIMARY KEY VALIDATION
# ============================================================

primary_keys = {
    "KNA1": ["Customer ID"],
    "LFA1": ["Vendor Number"],
    "VBAK": ["Sales Document"],
    "VBAP": ["Sales Document", "Item Number"],
    "LIKP": ["Delivery Number"],
    "LIPS": ["Delivery Number", "Item Number"],
    "VTTK": ["Shipment Number"],
    "VTTP": ["Shipment Number", "Item Number"]
}

print("=" * 90)
print("PRIMARY KEY VALIDATION")
print("=" * 90)

pk_results = []

for table_name, key_columns in primary_keys.items():

    df = tables[table_name]

    # Check whether PK columns exist
    missing_columns = [
        col for col in key_columns
        if col not in df.columns
    ]

    if missing_columns:
        print(f"\n{table_name}")
        print(f"Missing PK columns: {missing_columns}")
        continue

    # Check NULL values
    null_count = df[key_columns].isnull().any(axis=1).sum()

    # Check duplicate key combinations
    duplicate_count = df.duplicated(
        subset=key_columns,
        keep=False
    ).sum()

    # Count unique key combinations
    unique_key_count = df[key_columns].drop_duplicates().shape[0]

    # Total rows
    total_rows = len(df)

    # Determine result
    if null_count == 0 and duplicate_count == 0:
        status = "VALID"
    else:
        status = "CHECK REQUIRED"

    pk_results.append({
        "Table": table_name,
        "Primary Key": " + ".join(key_columns),
        "Total Rows": total_rows,
        "Unique Keys": unique_key_count,
        "NULL Key Rows": null_count,
        "Duplicate Key Rows": duplicate_count,
        "Status": status
    })

    print("\n" + "-" * 90)
    print(f"Table: {table_name}")
    print(f"Primary Key: {' + '.join(key_columns)}")
    print(f"Total Rows: {total_rows}")
    print(f"Unique Keys: {unique_key_count}")
    print(f"NULL Key Rows: {null_count}")
    print(f"Duplicate Key Rows: {duplicate_count}")
    print(f"Status: {status}")

print("\n" + "=" * 90)
print("PRIMARY KEY VALIDATION SUMMARY")
print("=" * 90)

pk_summary = pd.DataFrame(pk_results)
display(pk_summary)
```

**Step 19.2 — Check duplicate key combinations**

```python
# ============================================================
# DUPLICATE PRIMARY KEY RECORDS
# ============================================================

print("=" * 90)
print("DUPLICATE PRIMARY KEY RECORD CHECK")
print("=" * 90)

for table_name, key_columns in primary_keys.items():

    df = tables[table_name]

    duplicate_rows = df[
        df.duplicated(
            subset=key_columns,
            keep=False
        )
    ].sort_values(key_columns)

    print("\n" + "-" * 90)
    print(f"{table_name} | Key: {' + '.join(key_columns)}")

    if duplicate_rows.empty:
        print("No duplicate primary key combinations found.")
    else:
        display(duplicate_rows)
```

**Step 20.1 — Customer ID Referential Integrity**

```python
# ============================================================
# PHASE 2 - STEP 20.1
# CUSTOMER ID REFERENTIAL INTEGRITY
# ============================================================

print("=" * 90)
print("CUSTOMER ID REFERENTIAL INTEGRITY CHECK")
print("=" * 90)

customer_master = set(kna1["Customer ID"].dropna().unique())

customer_relationships = {
    "VBAK": "Customer ID",
    "LIKP": "Customer ID",
    "LIPS": "Customer ID",
    "VTTK": "Customer ID",
    "VTTP": "Customer ID"
}

customer_fk_results = []

for table_name, column in customer_relationships.items():

    df = tables[table_name]

    invalid_values = df[
        ~df[column].isin(customer_master)
    ][column].dropna().unique()

    customer_fk_results.append({
        "Child Table": table_name,
        "Foreign Key": column,
        "Parent Table": "KNA1",
        "Parent Key": "Customer ID",
        "Child Rows": len(df),
        "Invalid FK Values": len(invalid_values),
        "Status": "VALID" if len(invalid_values) == 0 else "CHECK REQUIRED"
    })

    print("\n" + "-" * 90)
    print(f"Child Table: {table_name}")
    print(f"Foreign Key: {column}")
    print(f"Parent Table: KNA1")
    print(f"Invalid Customer IDs: {len(invalid_values)}")

    if len(invalid_values) == 0:
        print("Status: VALID")
    else:
        print("Status: CHECK REQUIRED")
        print("Invalid values:", invalid_values)

customer_fk_summary = pd.DataFrame(customer_fk_results)

print("\n" + "=" * 90)
print("CUSTOMER REFERENTIAL INTEGRITY SUMMARY")
print("=" * 90)

display(customer_fk_summary)
```

**Step 21.1 — Define Header–Item Relationships**

```python
print("=" * 90)
print("STEP 21.1 — HEADER–ITEM RELATIONSHIP VALIDATION")
print("=" * 90)

header_item_relationships = {
    "VBAK": {
        "header_key": "Sales Document",
        "item_table": "VBAP",
        "item_key": "Sales Document"
    },
    "LIKP": {
        "header_key": "Delivery Number",
        "item_table": "LIPS",
        "item_key": "Delivery Number"
    },
    "VTTK": {
        "header_key": "Shipment Number",
        "item_table": "VTTP",
        "item_key": "Shipment Number"
    }
}

header_item_results = []

for header_table, relationship in header_item_relationships.items():

    header_df = tables[header_table]
    item_df = tables[relationship["item_table"]]

    header_key = relationship["header_key"]
    item_key = relationship["item_key"]

    # Check whether every item has a valid header
    invalid_items = item_df[
        ~item_df[item_key].isin(header_df[header_key])
    ]

    # Check whether every header has at least one item
    headers_without_items = header_df[
        ~header_df[header_key].isin(item_df[item_key])
    ]

    total_headers = len(header_df)
    total_items = len(item_df)

    valid_items = total_items - len(invalid_items)

    status = "VALID" if len(invalid_items) == 0 else "CHECK REQUIRED"

    print("\n" + "-" * 90)
    print(f"Header Table : {header_table}")
    print(f"Item Table   : {relationship['item_table']}")
    print(f"Header Key   : {header_key}")
    print(f"Item Key     : {item_key}")
    print(f"Total Headers: {total_headers}")
    print(f"Total Items  : {total_items}")
    print(f"Items with Valid Header: {valid_items}")
    print(f"Items with Invalid Header: {len(invalid_items)}")
    print(f"Headers Without Items: {len(headers_without_items)}")
    print(f"Status: {status}")

    header_item_results.append({
        "Header Table": header_table,
        "Item Table": relationship["item_table"],
        "Header Key": header_key,
        "Item Key": item_key,
        "Total Headers": total_headers,
        "Total Items": total_items,
        "Items with Invalid Header": len(invalid_items),
        "Headers Without Items": len(headers_without_items),
        "Status": status
    })

header_item_summary = pd.DataFrame(header_item_results)

print("\n" + "=" * 90)
print("HEADER–ITEM RELATIONSHIP SUMMARY")
print("=" * 90)

display(header_item_summary)
```

**Step 21.2 — Item-Level Key Validation**

```python
print("=" * 90)
print("STEP 21.2 — ITEM-LEVEL COMPOSITE KEY VALIDATION")
print("=" * 90)

item_keys = {
    "VBAP": ["Sales Document", "Item Number"],
    "LIPS": ["Delivery Number", "Item Number"],
    "VTTP": ["Shipment Number", "Item Number"]
}

item_key_results = []

for table_name, key_columns in item_keys.items():

    df = tables[table_name]

    # Check for null values in key columns
    null_key_rows = df[key_columns].isnull().any(axis=1).sum()

    # Check for duplicate composite keys
    duplicate_key_rows = df.duplicated(
        subset=key_columns,
        keep=False
    ).sum()

    # Count unique composite keys
    unique_key_count = df[key_columns].drop_duplicates().shape[0]

    total_rows = len(df)

    status = (
        "VALID"
        if null_key_rows == 0 and duplicate_key_rows == 0
        else "CHECK REQUIRED"
    )

    print("\n" + "-" * 90)
    print(f"Table: {table_name}")
    print(f"Composite Key: {' + '.join(key_columns)}")
    print(f"Total Rows: {total_rows}")
    print(f"Unique Composite Keys: {unique_key_count}")
    print(f"Rows with Null Key Values: {null_key_rows}")
    print(f"Rows with Duplicate Composite Keys: {duplicate_key_rows}")
    print(f"Status: {status}")

    item_key_results.append({
        "Table": table_name,
        "Composite Key": " + ".join(key_columns),
        "Total Rows": total_rows,
        "Unique Composite Keys": unique_key_count,
        "Null Key Rows": null_key_rows,
        "Duplicate Key Rows": duplicate_key_rows,
        "Status": status
    })

item_key_summary = pd.DataFrame(item_key_results)

print("\n" + "=" * 90)
print("ITEM-LEVEL COMPOSITE KEY SUMMARY")
print("=" * 90)

display(item_key_summary)
```

**Step 21.3 — Item Quantity Consistency Validation**

```python
print("=" * 90)
print("STEP 21.3 — ITEM QUANTITY CONSISTENCY VALIDATION")
print("=" * 90)

# ---------------------------------------------------------
# 1. Check Delivered Quantity against Ordered Quantity
# ---------------------------------------------------------

order_delivery_check = pd.merge(
    vbap[
        ["Sales Document", "Item Number", "Quantity"]
    ],
    lips[
        ["Sales Document", "Sales Item", "Delivered Quantity"]
    ],
    left_on=["Sales Document", "Item Number"],
    right_on=["Sales Document", "Sales Item"],
    how="left"
)

order_delivery_check["Quantity Difference"] = (
    order_delivery_check["Quantity"]
    - order_delivery_check["Delivered Quantity"]
)

invalid_delivery_quantity = order_delivery_check[
    order_delivery_check["Delivered Quantity"].notna()
    &
    (
        order_delivery_check["Delivered Quantity"]
        > order_delivery_check["Quantity"]
    )
]

# ---------------------------------------------------------
# 2. Check Shipped Quantity against Ordered Quantity
# ---------------------------------------------------------

order_shipment_check = pd.merge(
    vbap[
        ["Sales Document", "Item Number", "Quantity"]
    ],
    vttp[
        ["Sales Document", "Sales Item", "Shipped Quantity"]
    ],
    left_on=["Sales Document", "Item Number"],
    right_on=["Sales Document", "Sales Item"],
    how="left"
)

order_shipment_check["Quantity Difference"] = (
    order_shipment_check["Quantity"]
    - order_shipment_check["Shipped Quantity"]
)

invalid_shipment_quantity = order_shipment_check[
    order_shipment_check["Shipped Quantity"].notna()
    &
    (
        order_shipment_check["Shipped Quantity"]
        > order_shipment_check["Quantity"]
    )
]

# ---------------------------------------------------------
# 3. Display Results
# ---------------------------------------------------------

print("\n" + "-" * 90)
print("ORDERED vs DELIVERED QUANTITY")
print("-" * 90)

print("Order Items Checked:", len(order_delivery_check))
print(
    "Items with Delivered Quantity greater than Ordered Quantity:",
    len(invalid_delivery_quantity)
)

if len(invalid_delivery_quantity) == 0:
    print("Status: VALID")
else:
    print("Status: CHECK REQUIRED")
    display(invalid_delivery_quantity)

print("\n" + "-" * 90)
print("ORDERED vs SHIPPED QUANTITY")
print("-" * 90)

print("Order Items Checked:", len(order_shipment_check))
print(
    "Items with Shipped Quantity greater than Ordered Quantity:",
    len(invalid_shipment_quantity)
)

if len(invalid_shipment_quantity) == 0:
    print("Status: VALID")
else:
    print("Status: CHECK REQUIRED")
    display(invalid_shipment_quantity)

# ---------------------------------------------------------
# 4. Overall Status
# ---------------------------------------------------------

overall_status = (
    "VALID"
    if len(invalid_delivery_quantity) == 0
    and len(invalid_shipment_quantity) == 0
    else "CHECK REQUIRED"
)

print("\n" + "=" * 90)
print("STEP 21.3 OVERALL STATUS:", overall_status)
print("=" * 90)
```

**Step 22.7 — Validate Shipment Date Against Delivery Date**

```python
print("=" * 90)
print("STEP 22.7 — SHIPMENT DATE vs DELIVERY DATE VALIDATION")
print("=" * 90)

# ---------------------------------------------------------
# 1. Select required fields from VTTK and LIKP
# ---------------------------------------------------------

vttk_dates = vttk[
    [
        "Shipment Number",
        "Delivery Number",
        "Sales Document",
        "Shipment Date"
    ]
].copy()

likp_dates = likp[
    [
        "Delivery Number",
        "Sales Document",
        "Delivery Date"
    ]
].copy()

# ---------------------------------------------------------
# 2. Merge shipment and delivery information
#    using Delivery Number
# ---------------------------------------------------------

shipment_delivery_check = vttk_dates.merge(
    likp_dates,
    on="Delivery Number",
    how="left",
    suffixes=("_VTTK", "_LIKP")
)

# ---------------------------------------------------------
# 3. Identify shipments without a delivery header
# ---------------------------------------------------------

missing_delivery = shipment_delivery_check[
    shipment_delivery_check["Delivery Date"].isna()
]

# ---------------------------------------------------------
# 4. Check if shipment date is AFTER delivery date
# ---------------------------------------------------------

shipment_after_delivery = shipment_delivery_check[
    shipment_delivery_check["Delivery Date"].notna()
    &
    (
        shipment_delivery_check["Shipment Date"]
        >
        shipment_delivery_check["Delivery Date"]
    )
]

# ---------------------------------------------------------
# 5. Check if shipment date equals delivery date
# ---------------------------------------------------------

same_day = shipment_delivery_check[
    shipment_delivery_check["Delivery Date"].notna()
    &
    (
        shipment_delivery_check["Shipment Date"]
        ==
        shipment_delivery_check["Delivery Date"]
    )
]

# ---------------------------------------------------------
# 6. Check valid shipment-before-delivery records
# ---------------------------------------------------------

valid_sequence = shipment_delivery_check[
    shipment_delivery_check["Delivery Date"].notna()
    &
    (
        shipment_delivery_check["Shipment Date"]
        <=
        shipment_delivery_check["Delivery Date"]
    )
]

# ---------------------------------------------------------
# 7. Calculate shipment-to-delivery duration
# ---------------------------------------------------------

shipment_delivery_check["Shipment to Delivery Days"] = (
    shipment_delivery_check["Delivery Date"]
    - shipment_delivery_check["Shipment Date"]
).dt.days

# ---------------------------------------------------------
# 8. Check for negative duration
# ---------------------------------------------------------

negative_duration = shipment_delivery_check[
    shipment_delivery_check["Shipment to Delivery Days"] < 0
]

# ---------------------------------------------------------
# 9. Display validation results
# ---------------------------------------------------------

print("\n" + "-" * 90)
print("SHIPMENT → DELIVERY SEQUENCE VALIDATION")
print("-" * 90)

print(
    "Shipment Records Checked:",
    len(shipment_delivery_check)
)

print(
    "Valid Shipment ≤ Delivery Sequence:",
    len(valid_sequence)
)

print(
    "Same-Day Shipment and Delivery:",
    len(same_day)
)

print(
    "Shipment After Delivery:",
    len(shipment_after_delivery)
)

print(
    "Shipments Without Delivery Record:",
    len(missing_delivery)
)

print(
    "Negative Shipment-to-Delivery Durations:",
    len(negative_duration)
)

# ---------------------------------------------------------
# 10. Display invalid sequence records
# ---------------------------------------------------------

if len(shipment_after_delivery) > 0:
    print("\n" + "-" * 90)
    print("SHIPMENT AFTER DELIVERY — INVALID RECORDS")
    print("-" * 90)

    display(
        shipment_after_delivery[
            [
                "Shipment Number",
                "Delivery Number",
                "Sales Document_VTTK",
                "Shipment Date",
                "Delivery Date"
            ]
        ]
    )

# ---------------------------------------------------------
# 11. Display shipments without delivery
# ---------------------------------------------------------

if len(missing_delivery) > 0:
    print("\n" + "-" * 90)
    print("SHIPMENTS WITHOUT DELIVERY HEADER")
    print("-" * 90)

    display(
        missing_delivery[
            [
                "Shipment Number",
                "Delivery Number",
                "Sales Document_VTTK",
                "Shipment Date"
            ]
        ]
    )

# ---------------------------------------------------------
# 12. Shipment-to-delivery duration summary
# ---------------------------------------------------------

valid_durations = shipment_delivery_check[
    shipment_delivery_check["Shipment to Delivery Days"].notna()
]["Shipment to Delivery Days"]

print("\n" + "-" * 90)
print("SHIPMENT → DELIVERY DURATION SUMMARY")
print("-" * 90)

if len(valid_durations) > 0:
    print(
        "Minimum Shipment-to-Delivery Duration:",
        valid_durations.min(),
        "days"
    )

    print(
        "Maximum Shipment-to-Delivery Duration:",
        valid_durations.max(),
        "days"
    )

    print(
        "Average Shipment-to-Delivery Duration:",
        round(valid_durations.mean(), 2),
        "days"
    )

else:
    print("No valid shipment-to-delivery durations available.")

# ---------------------------------------------------------
# 13. Overall validation status
# ---------------------------------------------------------

if (
    len(shipment_after_delivery) == 0
    and len(negative_duration) == 0
    and len(missing_delivery) == 0
):
    overall_status = "VALID"
elif (
    len(shipment_after_delivery) == 0
    and len(negative_duration) == 0
):
    overall_status = "VALID WITH COMPLETENESS EXCEPTIONS"
else:
    overall_status = "CHECK REQUIRED"

print("\n" + "=" * 90)
print("STEP 22.7 OVERALL STATUS:", overall_status)
print("=" * 90)
```

```python
print("=" * 90)
print("23.5 — CASE CONSISTENCY VALIDATION")
print("=" * 90)

case_issues = []

for table_name, columns in categorical_fields.items():

    df = tables[table_name]

    for column in columns:

        series = df[column].dropna().astype(str)

        original_values = set(series.unique())

        normalized_values = {}

        for value in original_values:

            key = value.strip().lower()

            normalized_values.setdefault(key, []).append(value)

        inconsistent_groups = {
            key: values
            for key, values in normalized_values.items()
            if len(values) > 1
        }

        if len(inconsistent_groups) > 0:

            case_issues.append({
                "Table": table_name,
                "Column": column,
                "Inconsistent Groups": str(inconsistent_groups),
                "Status": "CHECK REQUIRED"
            })

if len(case_issues) > 0:

    case_issues_df = pd.DataFrame(case_issues)
    display(case_issues_df)

else:

    print("No capitalization inconsistencies found.")
```

**26.1 — Order-to-Delivery Cycle Duration Validation**

```python
print("=" * 90)
print("26.1 — ORDER-TO-DELIVERY CYCLE DURATION VALIDATION")
print("=" * 90)

orders = tables["VBAK"].copy()
deliveries = tables["LIKP"].copy()

# Convert dates
orders["Order Date"] = pd.to_datetime(
    orders["Order Date"],
    errors="coerce"
)

deliveries["Delivery Date"] = pd.to_datetime(
    deliveries["Delivery Date"],
    errors="coerce"
)

# Keep required columns
order_dates = orders[
    ["Sales Document", "Order Date"]
].drop_duplicates()

delivery_dates = deliveries[
    ["Sales Document", "Delivery Date"]
].drop_duplicates()

# Join Order Date with Delivery Date
cycle_check = order_dates.merge(
    delivery_dates,
    on="Sales Document",
    how="left"
)

# Calculate derived duration
cycle_check["Calculated Order-to-Delivery Days"] = (
    cycle_check["Delivery Date"] -
    cycle_check["Order Date"]
).dt.days

# Validate against expected conditions
negative_duration = cycle_check[
    cycle_check["Calculated Order-to-Delivery Days"] < 0
]

zero_duration = cycle_check[
    cycle_check["Calculated Order-to-Delivery Days"] == 0
]

missing_delivery = cycle_check[
    cycle_check["Delivery Date"].isna()
]

valid_duration = cycle_check[
    cycle_check["Calculated Order-to-Delivery Days"].notna()
]

print("Sales documents checked:", len(cycle_check))
print("Valid calculated durations:", len(valid_duration))
print("Missing delivery dates:", len(missing_delivery))
print("Negative durations:", len(negative_duration))
print("Zero-day durations:", len(zero_duration))

if len(negative_duration) == 0:
    print("Calculation Logic: VALID")
else:
    print("Calculation Logic: CHECK REQUIRED")

print("\nOrder-to-Delivery Duration Summary:")

if len(valid_duration) > 0:
    print(
        "Minimum days:",
        valid_duration["Calculated Order-to-Delivery Days"].min()
    )
    print(
        "Maximum days:",
        valid_duration["Calculated Order-to-Delivery Days"].max()
    )
    print(
        "Average days:",
        round(
            valid_duration[
                "Calculated Order-to-Delivery Days"
            ].mean(),
            2
        )
    )

print("\nSample calculated records:")
print(
    cycle_check[
        [
            "Sales Document",
            "Order Date",
            "Delivery Date",
            "Calculated Order-to-Delivery Days"
        ]
    ].head(10).to_string(index=False)
)
```

**Step 26.4 — Fulfillment Status Derivation Validation**

```python
print("=" * 90)
print("26.4 — FULFILLMENT STATUS DERIVATION VALIDATION")
print("=" * 90)

orders = tables["VBAK"].copy()
shipments = tables["VTTK"].copy()
deliveries = tables["LIKP"].copy()

# Clean keys
orders["Sales Document Clean"] = (
    orders["Sales Document"].astype(str).str.strip()
)

shipments["Sales Document Clean"] = (
    shipments["Sales Document"].astype(str).str.strip()
)

deliveries["Sales Document Clean"] = (
    deliveries["Sales Document"].astype(str).str.strip()
)

deliveries["Delivery Status Clean"] = (
    deliveries["Delivery Status"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# Create lookup sets
shipment_docs = set(
    shipments["Sales Document Clean"]
)

delivery_docs = set(
    deliveries["Sales Document Clean"]
)

delivered_docs = set(
    deliveries.loc[
        deliveries["Delivery Status Clean"] == "delivered",
        "Sales Document Clean"
    ]
)

# Derive fulfillment status
def derive_fulfillment_status(sales_document):

    if sales_document in delivered_docs:
        return "Delivered"

    elif sales_document in delivery_docs:
        return "In Delivery"

    elif sales_document in shipment_docs:
        return "Shipped"

    else:
        return "Order Processing"


orders["Derived Fulfillment Status"] = (
    orders["Sales Document Clean"]
    .apply(derive_fulfillment_status)
)

# Display distribution
print("\nDerived Fulfillment Status Distribution:")
print(
    orders["Derived Fulfillment Status"]
    .value_counts()
    .sort_index()
)

# Basic validation
print("\nTotal orders checked:", len(orders))
print(
    "Orders with derived status:",
    orders["Derived Fulfillment Status"].notna().sum()
)

if (
    orders["Derived Fulfillment Status"].notna().all()
):
    print("Derivation Logic: VALID")
else:
    print("Derivation Logic: CHECK REQUIRED")

# Show results
print("\nSample derived records:")
print(
    orders[
        [
            "Sales Document",
            "Order Status",
            "Derived Fulfillment Status"
        ]
    ].to_string(index=False)
)
```

```python
print("=" * 90)
print("27.1 — FINAL ETL READINESS: SOURCE TABLE CHECK")
print("=" * 90)

expected_tables = [
    "KNA1",
    "LFA1",
    "VBAK",
    "VBAP",
    "LIKP",
    "LIPS",
    "VTTK",
    "VTTP"
]

expected_counts = {
    "KNA1": 30,
    "LFA1": 30,
    "VBAK": 22,
    "VBAP": 23,
    "LIKP": 20,
    "LIPS": 21,
    "VTTK": 20,
    "VTTP": 21
}

missing_tables = []
count_mismatches = []

for table_name in expected_tables:

    if table_name not in tables:
        missing_tables.append(table_name)
        continue

    actual_count = len(tables[table_name])
    expected_count = expected_counts[table_name]

    if actual_count != expected_count:
        count_mismatches.append(
            {
                "Table": table_name,
                "Expected": expected_count,
                "Actual": actual_count
            }
        )

print("Expected tables:", len(expected_tables))
print("Available tables:", len(
    [t for t in expected_tables if t in tables]
))

print("\nMissing tables:", len(missing_tables))

if missing_tables:
    print(missing_tables)

print("\nRecord count mismatches:", len(count_mismatches))

if count_mismatches:
    print(pd.DataFrame(count_mismatches).to_string(index=False))

if len(missing_tables) == 0 and len(count_mismatches) == 0:
    print("\nSource Table Check: VALID")
else:
    print("\nSource Table Check: CHECK REQUIRED")
```

**27.2 — Final Mandatory Field & Key Readiness Check**

```python
print("=" * 90)
print("27.2 — FINAL ETL READINESS: MANDATORY FIELD & KEY CHECK")
print("=" * 90)

mandatory_fields = {
    "KNA1": [
        "Customer ID",
        "Customer Name"
    ],
    "LFA1": [
        "Vendor Number",
        "Vendor Name"
    ],
    "VBAK": [
        "Sales Document",
        "Order Date",
        "Customer ID"
    ],
    "VBAP": [
        "Sales Document",
        "Item Number",
        "Material Number",
        "Quantity",
        "Net Price"
    ],
    "LIKP": [
        "Delivery Number",
        "Delivery Date",
        "Sales Document"
    ],
    "LIPS": [
        "Delivery Number",
        "Item Number",
        "Material Number",
        "Delivered Quantity"
    ],
    "VTTK": [
        "Shipment Number",
        "Shipment Date",
        "Sales Document",
        "Delivery Number"
    ],
    "VTTP": [
        "Shipment Number",
        "Item Number",
        "Material Number",
        "Shipped Quantity"
    ]
}

missing_columns = []
missing_values = []

for table_name, columns in mandatory_fields.items():

    df = tables[table_name]

    for column in columns:

        # Check whether column exists
        if column not in df.columns:
            missing_columns.append(
                {
                    "Table": table_name,
                    "Column": column
                }
            )
            continue

        # Check missing values
        null_count = df[column].isna().sum()

        # Also check blank strings
        blank_count = (
            df[column]
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )

        total_missing = null_count + blank_count

        if total_missing > 0:
            missing_values.append(
                {
                    "Table": table_name,
                    "Column": column,
                    "Missing/Blank": total_missing
                }
            )

print("Mandatory fields checked:",
      sum(len(cols) for cols in mandatory_fields.values()))

print("\nMissing mandatory columns:",
      len(missing_columns))

if missing_columns:
    print(pd.DataFrame(missing_columns).to_string(index=False))

print("\nMandatory fields with missing/blank values:",
      len(missing_values))

if missing_values:
    print(pd.DataFrame(missing_values).to_string(index=False))

if len(missing_columns) == 0 and len(missing_values) == 0:
    print("\nMandatory Field Check: VALID")
else:
    print("\nMandatory Field Check: CHECK REQUIRED")
```

**27.3 — Final Primary Key & Referential Integrity Readiness Check**

```python
print("=" * 90)
print("27.3 — FINAL ETL READINESS: PRIMARY KEY & REFERENTIAL INTEGRITY CHECK")
print("=" * 90)

# ---------------------------------------------------------
# PRIMARY KEY CHECK
# ---------------------------------------------------------

primary_keys = {
    "KNA1": ["Customer ID"],
    "LFA1": ["Vendor Number"],
    "VBAK": ["Sales Document"],
    "VBAP": ["Sales Document", "Item Number"],
    "LIKP": ["Delivery Number"],
    "LIPS": ["Delivery Number", "Item Number"],
    "VTTK": ["Shipment Number"],
    "VTTP": ["Shipment Number", "Item Number"]
}

pk_issues = []

for table_name, key_columns in primary_keys.items():

    df = tables[table_name].copy()

    # Check NULL values
    null_count = df[key_columns].isna().any(axis=1).sum()

    # Check duplicate composite/single keys
    duplicate_count = df.duplicated(
        subset=key_columns,
        keep=False
    ).sum()

    if null_count > 0 or duplicate_count > 0:
        pk_issues.append(
            {
                "Table": table_name,
                "Key": " + ".join(key_columns),
                "Null Records": null_count,
                "Duplicate Records": duplicate_count
            }
        )

print("Primary key definitions checked:", len(primary_keys))
print("Tables with primary key issues:", len(pk_issues))

if pk_issues:
    print(pd.DataFrame(pk_issues).to_string(index=False))


# ---------------------------------------------------------
# FOREIGN KEY CHECK
# ---------------------------------------------------------

relationships = [
    ("KNA1", "Customer ID", "VBAK", "Customer ID"),
    ("KNA1", "Customer ID", "LIKP", "Customer ID"),
    ("KNA1", "Customer ID", "LIPS", "Customer ID"),
    ("KNA1", "Customer ID", "VTTK", "Customer ID"),
    ("KNA1", "Customer ID", "VTTP", "Customer ID"),

    ("VBAK", "Sales Document", "VBAP", "Sales Document"),
    ("VBAK", "Sales Document", "LIKP", "Sales Document"),
    ("VBAK", "Sales Document", "LIPS", "Sales Document"),
    ("VBAK", "Sales Document", "VTTK", "Sales Document"),
    ("VBAK", "Sales Document", "VTTP", "Sales Document"),

    ("LIKP", "Delivery Number", "LIPS", "Delivery Number"),
    ("LIKP", "Delivery Number", "VTTK", "Delivery Number"),
    ("LIKP", "Delivery Number", "VTTP", "Delivery Number"),

    ("VTTK", "Shipment Number", "VTTP", "Shipment Number")
]

fk_issues = []

for parent_table, parent_column, child_table, child_column in relationships:

    parent_keys = set(
        tables[parent_table][parent_column]
        .dropna()
        .astype(str)
        .str.strip()
    )

    child_keys = (
        tables[child_table][child_column]
        .dropna()
        .astype(str)
        .str.strip()
    )

    invalid_keys = child_keys[
        ~child_keys.isin(parent_keys)
    ]

    if len(invalid_keys) > 0:
        fk_issues.append(
            {
                "Parent": parent_table,
                "Parent Key": parent_column,
                "Child": child_table,
                "Child Key": child_column,
                "Invalid Records": len(invalid_keys)
            }
        )

print("\nForeign key relationships checked:", len(relationships))
print("Relationships with referential integrity issues:",
      len(fk_issues))

if fk_issues:
    print(pd.DataFrame(fk_issues).to_string(index=False))


# ---------------------------------------------------------
# OVERALL RESULT
# ---------------------------------------------------------

if len(pk_issues) == 0 and len(fk_issues) == 0:
    print("\nPrimary Key & Referential Integrity Check: VALID")
else:
    print("\nPrimary Key & Referential Integrity Check: CHECK REQUIRED")
```

**Step 27.4 — Final Date & Numeric Readiness Check**

```python
print("=" * 90)
print("27.4 — FINAL ETL READINESS: DATE & NUMERIC DATA CHECK")
print("=" * 90)

# ---------------------------------------------------------
# DATE FIELD CHECK
# ---------------------------------------------------------

date_columns = {
    "VBAK": ["Order Date"],
    "VBAP": ["Delivery Date"],
    "LIKP": ["Delivery Date"],
    "LIPS": ["Delivery Date"],
    "VTTK": ["Shipment Date"],
    "VTTP": ["Shipment Date"]
}

date_issues = []

for table_name, columns in date_columns.items():
    df = tables[table_name].copy()

    for col in columns:
        converted = pd.to_datetime(df[col], errors="coerce")

        invalid_count = converted.isna().sum()

        if invalid_count > 0:
            date_issues.append({
                "Table": table_name,
                "Column": col,
                "Invalid/Missing Dates": invalid_count
            })

print("Date fields checked:", sum(len(cols) for cols in date_columns.values()))
print("Date fields with issues:", len(date_issues))

if date_issues:
    print(pd.DataFrame(date_issues).to_string(index=False))


# ---------------------------------------------------------
# NUMERIC FIELD CHECK
# ---------------------------------------------------------

numeric_columns = {
    "VBAP": ["Quantity", "Net Price"],
    "LIPS": ["Delivered Quantity", "Net Price"],
    "VTTP": ["Shipped Quantity"]
}

numeric_issues = []

for table_name, columns in numeric_columns.items():
    df = tables[table_name].copy()

    for col in columns:
        numeric_values = pd.to_numeric(df[col], errors="coerce")

        missing_count = numeric_values.isna().sum()
        negative_count = (numeric_values < 0).sum()
        zero_count = (numeric_values == 0).sum()

        if missing_count > 0 or negative_count > 0 or zero_count > 0:
            numeric_issues.append({
                "Table": table_name,
                "Column": col,
                "Missing": missing_count,
                "Negative": negative_count,
                "Zero": zero_count
            })

print("\nNumeric fields checked:",
      sum(len(cols) for cols in numeric_columns.values()))
print("Numeric fields with issues:", len(numeric_issues))

if numeric_issues:
    print(pd.DataFrame(numeric_issues).to_string(index=False))


# ---------------------------------------------------------
# FINAL RESULT
# ---------------------------------------------------------

if len(date_issues) == 0 and len(numeric_issues) == 0:
    print("\nDate & Numeric Readiness Check: VALID")
else:
    print("\nDate & Numeric Readiness Check: CHECK REQUIRED")
```

**Step 31.9 — Build Delivery_Analytics**

```python
# ============================================================================
# STEP 31.9 — DELIVERY_ANALYTICS TRANSFORMATION
# ============================================================================

print("=" * 90)
print("STEP 31.9 — DELIVERY_ANALYTICS TRANSFORMATION")
print("=" * 90)

# --------------------------------------------------------------------------
# 1. Prepare Orders
# --------------------------------------------------------------------------

orders_analytics = orders[
    ["order_id", "customer_id", "order_date"]
].copy()

orders_analytics["order_date"] = pd.to_datetime(
    orders_analytics["order_date"],
    errors="coerce"
)


# --------------------------------------------------------------------------
# 2. Prepare Shipment Data
# --------------------------------------------------------------------------

shipments_analytics = shipments[
    ["order_id", "shipment_date"]
].copy()

shipments_analytics["shipment_date"] = pd.to_datetime(
    shipments_analytics["shipment_date"],
    errors="coerce"
)


# --------------------------------------------------------------------------
# 3. Prepare Delivery Data
#    IMPORTANT: Keep delivery_status for fulfillment derivation
# --------------------------------------------------------------------------

delivery_analytics_source = delivery_status[
    [
        "order_id",
        "delivery_date",
        "delivery_status"
    ]
].copy()

delivery_analytics_source["delivery_date"] = pd.to_datetime(
    delivery_analytics_source["delivery_date"],
    errors="coerce"
)


# --------------------------------------------------------------------------
# 4. Merge Orders + Shipments
# --------------------------------------------------------------------------

delivery_analytics = orders_analytics.merge(
    shipments_analytics,
    on="order_id",
    how="left"
)


# --------------------------------------------------------------------------
# 5. Merge Orders + Shipments + Delivery
# --------------------------------------------------------------------------

delivery_analytics = delivery_analytics.merge(
    delivery_analytics_source,
    on="order_id",
    how="left"
)


# --------------------------------------------------------------------------
# 6. Calculate Order → Shipment Duration
# --------------------------------------------------------------------------

delivery_analytics["order_to_ship_days"] = (
    delivery_analytics["shipment_date"]
    - delivery_analytics["order_date"]
).dt.days


# --------------------------------------------------------------------------
# 7. Calculate Order → Delivery Duration
# --------------------------------------------------------------------------

delivery_analytics["order_to_delivery_days"] = (
    delivery_analytics["delivery_date"]
    - delivery_analytics["order_date"]
).dt.days


# --------------------------------------------------------------------------
# 8. Calculate Shipment → Delivery Duration
# --------------------------------------------------------------------------

delivery_analytics["shipment_to_delivery_days"] = (
    delivery_analytics["delivery_date"]
    - delivery_analytics["shipment_date"]
).dt.days


# --------------------------------------------------------------------------
# 9. Derive Fulfillment Status
#
# Rules:
#   - Delivered      → Delivery record exists and Delivery Status = Delivered
#   - In Delivery    → Delivery record exists but not Delivered
#   - Shipped        → Shipment exists but no Delivery record
#   - Order Processing → No shipment and no delivery
# --------------------------------------------------------------------------

def derive_fulfillment_status(row):

    if pd.notna(row["delivery_status"]):

        if str(row["delivery_status"]).strip().lower() == "delivered":
            return "Delivered"

        else:
            return "In Delivery"

    elif pd.notna(row["shipment_date"]):
        return "Shipped"

    else:
        return "Order Processing"


delivery_analytics["fulfillment_status"] = (
    delivery_analytics.apply(
        derive_fulfillment_status,
        axis=1
    )
)


# --------------------------------------------------------------------------
# 10. Remove intermediate delivery_status column
# --------------------------------------------------------------------------

delivery_analytics = delivery_analytics[
    [
        "order_id",
        "customer_id",
        "order_date",
        "shipment_date",
        "delivery_date",
        "order_to_ship_days",
        "order_to_delivery_days",
        "shipment_to_delivery_days",
        "fulfillment_status"
    ]
]


# --------------------------------------------------------------------------
# 11. Validation
# --------------------------------------------------------------------------

print("\nTransformed records:", len(delivery_analytics))
print("Columns:", len(delivery_analytics.columns))

print("\nColumns:")
print(list(delivery_analytics.columns))

print("\nFulfillment Status:")
print(
    delivery_analytics["fulfillment_status"].value_counts()
)

print("\nMissing values:")
print(
    delivery_analytics.isnull().sum()
)

print("\nDuration statistics:")

print("\nOrder → Ship:")
print(
    delivery_analytics["order_to_ship_days"].describe()
)

print("\nOrder → Delivery:")
print(
    delivery_analytics["order_to_delivery_days"].describe()
)

print("\nShipment → Delivery:")
print(
    delivery_analytics["shipment_to_delivery_days"].describe()
)

print("\nPreview:")
display(delivery_analytics.head())

print("\nDelivery Analytics transformation: VALID")
```

**Step 31.10 — Create Date Dimension**

```python
# ============================================================
# STEP 31.10 — BUILD DATE_DIMENSION
# ============================================================

print("=" * 90)
print("31.10 — BUILD DATE_DIMENSION")
print("=" * 90)

# ------------------------------------------------------------
# Determine the complete date range
# ------------------------------------------------------------

date_sources = [
    pd.to_datetime(tables["VBAK"]["Order Date"]),
    pd.to_datetime(tables["VBAP"]["Delivery Date"]),
    pd.to_datetime(tables["LIKP"]["Delivery Date"]),
    pd.to_datetime(tables["VTTK"]["Shipment Date"])
]

start_date = min(series.min() for series in date_sources)
end_date = max(series.max() for series in date_sources)

print("Date range:")
print("Start:", start_date.date())
print("End  :", end_date.date())

# ------------------------------------------------------------
# Generate continuous date range
# ------------------------------------------------------------

date_range = pd.date_range(
    start=start_date,
    end=end_date,
    freq="D"
)

date_dimension = pd.DataFrame({
    "full_date": date_range
})

# ------------------------------------------------------------
# Create Date Dimension attributes
# ------------------------------------------------------------

date_dimension["date_key"] = (
    date_dimension["full_date"].dt.strftime("%Y%m%d").astype(int)
)

date_dimension["year"] = (
    date_dimension["full_date"].dt.year
)

date_dimension["month"] = (
    date_dimension["full_date"].dt.month
)

date_dimension["month_name"] = (
    date_dimension["full_date"].dt.strftime("%B")
)

date_dimension["quarter"] = (
    date_dimension["full_date"].dt.quarter
)

date_dimension["day"] = (
    date_dimension["full_date"].dt.day
)

date_dimension["day_name"] = (
    date_dimension["full_date"].dt.strftime("%A")
)

date_dimension["week_number"] = (
    date_dimension["full_date"].dt.isocalendar().week.astype(int)
)

# ------------------------------------------------------------
# Reorder columns to match MySQL
# ------------------------------------------------------------

date_dimension = date_dimension[
    [
        "date_key",
        "full_date",
        "year",
        "month",
        "month_name",
        "quarter",
        "day",
        "day_name",
        "week_number"
    ]
]

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\nDate Dimension preview:")
display(date_dimension.head())

print("\nDate Dimension last records:")
display(date_dimension.tail())

print("\nTotal date records:", len(date_dimension))

print("\nMissing values:")
print(date_dimension.isna().sum())

print(
    "\nDuplicate date keys:",
    date_dimension["date_key"].duplicated().sum()
)

print(
    "Duplicate dates:",
    date_dimension["full_date"].duplicated().sum()
)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

if (
    len(date_dimension) > 0
    and date_dimension["date_key"].notna().all()
    and date_dimension["full_date"].notna().all()
    and date_dimension["date_key"].duplicated().sum() == 0
    and date_dimension["full_date"].duplicated().sum() == 0
):
    print("\nDate_Dimension Transformation: VALID")
else:
    print("\nDate_Dimension Transformation: CHECK REQUIRED")
```

**Step 32 — Final Transformed Data Validation**

```python
# ============================================================
# STEP 32 — FINAL TRANSFORMED DATA VALIDATION
# ============================================================

print("=" * 90)
print("STEP 32 — FINAL TRANSFORMED DATA VALIDATION")
print("=" * 90)

# ------------------------------------------------------------
# 32.1 — Record Count Validation
# ------------------------------------------------------------

record_count_checks = {
    "Customers": (len(tables["KNA1"]), len(customers)),
    "Carriers": (len(tables["LFA1"]), len(carriers)),
    "Orders": (len(tables["VBAK"]), len(orders)),
    "Order_Items": (len(tables["VBAP"]), len(order_items)),
    "Delivery_Status": (len(tables["LIKP"]), len(delivery_status)),
    "Delivery_Items": (len(tables["LIPS"]), len(delivery_items)),
    "Shipments": (len(tables["VTTK"]), len(shipments)),
    "Shipment_Items": (len(tables["VTTP"]), len(shipment_items)),
    "Delivery_Analytics": (len(orders), len(analytics)),
    "Date_Dimension": (30, len(date_dimension))
}

print("\n32.1 — RECORD COUNT VALIDATION")
print("-" * 90)

record_count_issues = 0

for table_name, (source_count, transformed_count) in record_count_checks.items():

    status = "VALID" if source_count == transformed_count else "CHECK REQUIRED"

    if status != "VALID":
        record_count_issues += 1

    print(
        f"{table_name:<20} "
        f"Source: {source_count:<5} "
        f"Transformed: {transformed_count:<5} "
        f"{status}"
    )

# ------------------------------------------------------------
# 32.2 — Primary / Composite Key Validation
# ------------------------------------------------------------

print("\n32.2 — KEY VALIDATION")
print("-" * 90)

key_checks = {
    "Customers": (
        customers,
        ["customer_id"]
    ),
    "Carriers": (
        carriers,
        ["carrier_id"]
    ),
    "Orders": (
        orders,
        ["order_id"]
    ),
    "Order_Items": (
        order_items,
        ["order_id", "item_number"]
    ),
    "Delivery_Status": (
        delivery_status,
        ["delivery_id"]
    ),
    "Delivery_Items": (
        delivery_items,
        ["delivery_id", "item_number"]
    ),
    "Shipments": (
        shipments,
        ["shipment_id"]
    ),
    "Shipment_Items": (
        shipment_items,
        ["shipment_id", "item_number"]
    ),
    "Delivery_Analytics": (
        analytics,
        ["order_id"]
    ),
    "Date_Dimension": (
        date_dimension,
        ["date_key"]
    )
}

key_issues = 0

for table_name, (df, key_columns) in key_checks.items():

    duplicate_count = df.duplicated(
        subset=key_columns
    ).sum()

    null_count = df[key_columns].isna().sum().sum()

    if duplicate_count == 0 and null_count == 0:
        status = "VALID"
    else:
        status = "CHECK REQUIRED"
        key_issues += 1

    print(
        f"{table_name:<20} "
        f"Null keys: {null_count:<3} "
        f"Duplicate keys: {duplicate_count:<3} "
        f"{status}"
    )

# ------------------------------------------------------------
# 32.3 — Mandatory Field Validation
# ------------------------------------------------------------

print("\n32.3 — MANDATORY FIELD VALIDATION")
print("-" * 90)

mandatory_fields = {
    "Customers": (
        customers,
        ["customer_id", "customer_name"]
    ),
    "Carriers": (
        carriers,
        ["carrier_id", "carrier_name"]
    ),
    "Orders": (
        orders,
        ["order_id", "customer_id", "order_date"]
    ),
    "Order_Items": (
        order_items,
        [
            "order_id",
            "item_number",
            "product_id",
            "order_quantity",
            "unit_price"
        ]
    ),
    "Delivery_Status": (
        delivery_status,
        [
            "delivery_id",
            "order_id",
            "customer_id",
            "delivery_date"
        ]
    ),
    "Delivery_Items": (
        delivery_items,
        [
            "delivery_id",
            "item_number",
            "product_id",
            "delivered_quantity",
            "net_price"
        ]
    ),
    "Shipments": (
        shipments,
        [
            "shipment_id",
            "order_id",
            "delivery_id",
            "customer_id",
            "shipment_date"
        ]
    ),
    "Shipment_Items": (
        shipment_items,
        [
            "shipment_id",
            "item_number",
            "product_id",
            "shipped_quantity"
        ]
    ),
    "Delivery_Analytics": (
        analytics,
        [
            "order_id",
            "customer_id",
            "order_date",
            "fulfillment_status"
        ]
    ),
    "Date_Dimension": (
        date_dimension,
        [
            "date_key",
            "full_date",
            "year",
            "month",
            "quarter"
        ]
    )
}

mandatory_issues = 0

for table_name, (df, columns) in mandatory_fields.items():

    missing_count = df[columns].isna().sum().sum()

    if missing_count == 0:
        status = "VALID"
    else:
        status = "CHECK REQUIRED"
        mandatory_issues += 1

    print(
        f"{table_name:<20} "
        f"Missing mandatory values: {missing_count:<3} "
        f"{status}"
    )

# ------------------------------------------------------------
# 32.4 — Referential Integrity After Transformation
# ------------------------------------------------------------

print("\n32.4 — REFERENTIAL INTEGRITY VALIDATION")
print("-" * 90)

fk_issues = 0

# Orders → Customers
invalid = (~orders["customer_id"].isin(
    customers["customer_id"]
)).sum()

print(
    f"Orders → Customers: "
    f"{invalid} invalid customer IDs"
)

fk_issues += invalid

# Order_Items → Orders
invalid = (~order_items["order_id"].isin(
    orders["order_id"]
)).sum()

print(
    f"Order_Items → Orders: "
    f"{invalid} invalid order IDs"
)

fk_issues += invalid

# Delivery_Status → Orders
invalid = (~delivery_status["order_id"].isin(
    orders["order_id"]
)).sum()

print(
    f"Delivery_Status → Orders: "
    f"{invalid} invalid order IDs"
)

fk_issues += invalid

# Delivery_Items → Delivery_Status
invalid = (~delivery_items["delivery_id"].isin(
    delivery_status["delivery_id"]
)).sum()

print(
    f"Delivery_Items → Delivery_Status: "
    f"{invalid} invalid delivery IDs"
)

fk_issues += invalid

# Shipments → Orders
invalid = (~shipments["order_id"].isin(
    orders["order_id"]
)).sum()

print(
    f"Shipments → Orders: "
    f"{invalid} invalid order IDs"
)

fk_issues += invalid

# Shipments → Delivery_Status
invalid = (~shipments["delivery_id"].isin(
    delivery_status["delivery_id"]
)).sum()

print(
    f"Shipments → Delivery_Status: "
    f"{invalid} invalid delivery IDs"
)

fk_issues += invalid

# Shipment_Items → Shipments
invalid = (~shipment_items["shipment_id"].isin(
    shipments["shipment_id"]
)).sum()

print(
    f"Shipment_Items → Shipments: "
    f"{invalid} invalid shipment IDs"
)

fk_issues += invalid

# Delivery_Analytics → Orders
invalid = (~analytics["order_id"].isin(
    orders["order_id"]
)).sum()

print(
    f"Delivery_Analytics → Orders: "
    f"{invalid} invalid order IDs"
)

fk_issues += invalid

# ------------------------------------------------------------
# 32.5 — Final Result
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("STEP 32 — FINAL TRANSFORMED DATA VALIDATION RESULT")
print("=" * 90)

if (
    record_count_issues == 0
    and key_issues == 0
    and mandatory_issues == 0
    and fk_issues == 0
):
    print("\nFINAL TRANSFORMED DATA VALIDATION: VALID")
    print("All transformed datasets are ready for MySQL loading.")
else:
    print("\nFINAL TRANSFORMED DATA VALIDATION: CHECK REQUIRED")
    print("Review the validation results before MySQL loading.")
```
